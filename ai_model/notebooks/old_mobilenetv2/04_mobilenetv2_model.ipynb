```python
IMG_SIZE = (224, 224)
NUM_CLASSES = 101
```

```python
base_model = MobileNetV2(
    input_shape=(224,224,3),
    include_top=False,
    weights="imagenet"
)
```

```python
base_model.trainable = False

print("Base Model Trainable:", base_model.trainable)
```

```python
model = models.Sequential([
    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.3),

    layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )
])
```

```python
model.summary()
```
