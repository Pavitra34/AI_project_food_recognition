```python
import tensorflow as tf
import matplotlib.pyplot as plt
import pickle
```

```python
model = tf.keras.models.load_model("../saved_models/mobilenetv2_food_model.keras")
```

```python
with open("../saved_models/history.pkl", "rb") as f:
    history = pickle.load(f)

print("History Loaded Successfully")
```

```python
history.keys()
```

```python
plt.figure(figsize=(10,5))

plt.plot(history["accuracy"], label="Training Accuracy")
plt.plot(history["val_accuracy"], label="Validation Accuracy")

plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(history["loss"], label="Training Loss")
plt.plot(history["val_loss"], label="Validation Loss")

plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.show()
```

```python
print("Training Accuracy :", history["accuracy"][-1])
print("Validation Accuracy :", history["val_accuracy"][-1])
```
